## Mini glossary

- **Token** - a chunk of text (often 3-4 characters) that a model reads and bills by.
- **Latency** - how long a request takes, start to finish.
- **TTFT / TTFB** - time to first token / first byte: how long until the user sees *anything*.
- **Median** - the middle value after sorting. Ignores one freak slow run.
- **p95** - 95% of requests were faster than this. Measures the "bad but normal" requests.
- **async / await** - Python's way of doing other work while waiting on the network.
- **SSE (Server-Sent Events)** - a simple way for a server to push text to the browser piece by piece.
- **Middleware** - code that wraps every request, like a guard at the door who sees everyone going in and out.
- **Mock** - a fake stand-in for a real thing (an LLM, a database) so a test runs with no network.
- **BPE (byte-pair encoding)** - how tokenizers are built: keep merging the most common pairs of characters.
- **Temperature** - the "creativity dial": low = predictable, high = varied.
- **Context window** - the maximum number of tokens a model can look at at once.
- **Canonical key** - one standard form that all "equivalent" things share (every anagram of "eat" -> the same key).
- **Hashable** - can be a dict key / set member. Needs a hash that never changes, so mutable things (lists) aren't.
